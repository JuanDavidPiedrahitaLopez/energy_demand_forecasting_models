```python
date_close =  pd.read_csv("./csv/EC_GER.csv", delimiter=";")
# se escogen solo las columnas requeridas
date_close = date_close[["Start date", "grid load [MWh] Calculated resolutions"]]
date_close["Start date"] = pd.to_datetime(date_close["Start date"])
date_close = date_close.rename(columns={"grid load [MWh] Calculated resolutions": "value"})
date_close["value"] = date_close["value"].str.replace(",", "", regex=False).astype(float)
date_close.head()
```

```python
df_date_close = date_close.set_index("Start date")
df_date_close.index = pd.to_datetime(df_date_close.index)
df_weekly = df_date_close.resample('W-MON').mean()
#df_weekly = df_date_close.resample('W-MON').sum()
#df_weekly = df_weekly.reset_index()
df_weekly
```
